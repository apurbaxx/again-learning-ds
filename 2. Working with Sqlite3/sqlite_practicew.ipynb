```python
import _sqlite3
```

```python
connection = _sqlite3.connect("example1.db")
cursor = connection.cursor()
```

```python
cursor.execute('''
Create Table If Not Exists employees(
id Integer Primary Key,
name String Not Null,
department Text,
age Integer
)
''')
connection.commit()
```

```python
cursor.execute("select * from employees")
```

```python
cursor.execute('''
Insert into employees(name,age,department)
    VALUES("Apurba",18,"Engineer")
''')
connection.commit()
```

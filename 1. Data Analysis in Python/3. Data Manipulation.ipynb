```python
##Filling missing values with mean of the column
```

```python
df["Total_Revenue_fillNA"] = df["Total Revenue"].fillna(df["Total Revenue"].mean())
```

```python
df["Total_Revenue_fillNA"]
```

```python
df.dtypes
```

```python
df.rename(columns={'Date':'Sale Date'}, inplace=True)
```

```python
df.head(5)
```

```python
#change datatypes
df["Units Sold New"]=df["Units Sold"].fillna(df["Units Sold"].mean()).astype(float)
```

```python
df.head(5)
```

```python
df["Correct Units Sold"] = df["Units Sold"].apply(lambda x:x*2)
```

```python
df.head(10)
```

```python
#Data Aggregating and Grouping
group_mean = df.groupby(["Product Category"])["Units Sold"].mean()
```

```python
group_mean
```

```python
#aggregate multiple functions
group_agg = df.groupby(["Product Category"])["Units Sold"].agg("mean", "sum", "count")
```

```python
###Merging and joining dataframes

df1 = pd.DataFrame({"Key":["A","B","C"], "value1":[1,2,3]})
df2 = pd.DataFrame({"Key":["A","B","D"], "value2":[4,5,6]})
```

```python
df1
```

```python
df2
```

```python
pd.merge(df1,df2, on="Key", how="inner")
```

```python
pd.merge(df1,df2, on="Key",how="outer")
```

```python
pd.merge(df1,df2,on="Key",how="left")
```

```python
pd.merge(df1,df2,on="Key",how="right")
```

```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5]
y = [1,6,10,16,19]
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.title("Basic line plot")
plt.grid(True)
plt.plot(x,y, color="red", linestyle="dashed", marker="o", linewidth=2, markersize=8, markerfacecolor="blue")
```

```python
import pandas as pd
sales_data_df = pd.read_csv("sales_data.csv")
sales_data_df.head(5)
```

```python
sales_data_df.info()
```

```python
total_sales_by_product_sales = sales_data_df.groupby("Product Category")["Total Revenue"].sum()
print(total_sales_by_product_sales)
```

```python
total_sales_by_product_sales.plot(kind="bar", color="teal")
```

```python
##plot trend sales over time
sales_trend = sales_data_df.groupby("Date")["Total Revenue"].sum().reset_index()
plt.plot(sales_trend["Date"], sales_trend["Total Revenue"], color="red")
```

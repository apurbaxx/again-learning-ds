Logging is a crusical ascept of any application, providing a way to track events,erros and operational information. Python's built in logging module offers a flexible framework for emitting log messages from python programs.

```python
import logging
```

```python
##Configure the basic config settings

logging.basicConfig(
    level=logging.DEBUG,
    format="%(asctime)s - %(name)s - %(levelname)s - %(message)s",
    datefmt="%Y-%M-%D %H:%M:%S",
)

##Log messages with different severity levels

logging.debug("This is a debug message")
logging.info("This is an info message")
logging.warning("This is a warning message")
logging.error("This is an error message")
logging.critical("This is a critical message")
```

```python
import logging

logging.basicConfig(
    filename="app.log",
    filemode="w",
    level=logging.DEBUG,
    format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
    datefmt="%Y-%m-%d %H:%M:%S",
    force=True
)

logging.debug("This is a debug message")
logging.info("This is an info message")
logging.warning("This is a warning message")
logging.error("This is an error message")
logging.critical("This is a critical message")
```

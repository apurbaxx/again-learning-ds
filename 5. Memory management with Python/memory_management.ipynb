- Reference counting is the primary method python uses to manage memory. Each Object in python maintains a count of references pointing to it. With the reference count drops to zero, the memory occupied by the object is deallocated.

```python
import sys

a = []
print(sys.getrefcount(a))
# 2 (one reference comes from "a" and one from getrefcount())
```
